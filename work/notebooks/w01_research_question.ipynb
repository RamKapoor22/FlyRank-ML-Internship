## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess
import pandas as pd
import numpy as np

# Clone repo in Colab if not present and adjust working directory
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    if os.path.basename(os.getcwd()) != REPO_DIR:
        os.chdir(REPO_DIR)

# Load data
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. Percentage of declining pages
declining_count = (df["trend_direction"].str.lower() == "down").sum()
declining_pct = (declining_count / len(df)) * 100

# 2. Search volume vs impressions correlation
corr_val = df["search_volume"].corr(df["impressions_90d"])

# 3. Median age difference
median_age_down = df[df["trend_direction"] == "down"]["content_age_days"].median()
median_age_up = df[df["trend_direction"] == "up"]["content_age_days"].median()

print(f"1. Total Declining Pages: {declining_count} / {len(df)} ({declining_pct:.1f}%)")
print(f"2. Search Volume vs Impressions Correlation: {corr_val:.3f}")
print(f"3. Median Age - Down pages: {median_age_down:.0f} days | Up pages: {median_age_up:.0f} days")
```
